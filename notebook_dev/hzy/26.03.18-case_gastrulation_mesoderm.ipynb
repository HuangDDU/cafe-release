```python
import pandas as pd
import cafe

cafe.logger.setLevel("DEBUG")
```

```python
sub_filename = ".cafe/gastrulation_mesoderm/h5ad/gastrulation_mesoderm.h5ad"
fadata = cafe.data.read_h5ad(sub_filename)
fadata.prior_information["basis"]="X_umap"
fadata
```

```python
milestone_network = pd.DataFrame(
    [
        ["Nascent mesoderm", "Caudal Mesoderm"],
        ["Nascent mesoderm", "Intermediate mesoderm"],
        ["Nascent mesoderm", "Pharyngeal mesoderm"],
        ["Nascent mesoderm", "ExE mesoderm"],
        ["Caudal Mesoderm", "Paraxial mesoderm"],
        ["Paraxial mesoderm", "Somitic mesoderm"],
        ["Caudal Mesoderm", "Mixed mesoderm"],
        ["Caudal Mesoderm", "PGC"],
    ],
    columns=["from", "to"],
)

fadata.add_trajectory_mannually(milestone_network=milestone_network)
```

```python
cafe.plot.plot_trajectory(fadata)
cafe.plot.plot_graph(fadata)
```

```python
fadata.id = "gastrulation_mesoderm"
fadata.check_result_dir()
fadata.write_trajectory_dict(model_name_list=["ref"])
```
