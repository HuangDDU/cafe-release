# 26.03.11-plot_gene_trends

```python
import cafe
```

```python
fadata = cafe.data.read_pancreas() # pancrease gloden dataset from scvelo.
cafe.plot.plot_trajectory(fadata, show_milestone_labels=True)
fadata
```

preprocess for smoother curve

```python
import scvelo as scv

scv.pp.filter_and_normalize(fadata)
# scv.pp.moments(fadata)
```

## 0. linear trend without lineage

```python
cafe.plot.plot_gene_linear_trends(fadata, genes="Pdx1")
cafe.plot.plot_gene_linear_trends(fadata, genes="Pdx1", curve_mode="pygam")
```

## 1. curve mode: gaussian, pygam

```python
cafe.plot.plot_gene_trends(fadata, genes="Pdx1", show_cell=False)
cafe.plot.plot_gene_trends(fadata, genes="Pdx1", curve_mode="pygam", show_cell=False)
```

## 2. show cell.

```python
cafe.plot.plot_gene_trends(fadata, genes="Pdx1", show_cell=True)
cafe.plot.plot_gene_trends(fadata, genes="Pdx1", curve_mode="pygam", show_cell=True, save=True)
```

## 3. show multiple genes

```python
cafe.plot.plot_gene_trends(
    fadata,
    genes=["Arx", "Pdx1", "Hhex", "Enpp1"],
    curve_mode="pygam",
    n_cols=2,
    show_cell=True,
    save=True
)
```
