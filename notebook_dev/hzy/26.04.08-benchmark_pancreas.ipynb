```python
import cafe

# cafe.logger.setLevel("DEBUG")
img_dir = "img/26.04.08-benchmark_pancreas"
```

```python
fadata = cafe.data.read_pancreas()
cafe.plot.plot_trajectory(fadata) # waypoint wrapper
fadata
```

benchmark method run

```python
fadata = cafe.benchmark.run(fadata)
```

benchmark method caculation and normalization

```python
benchmark_df = cafe.benchmark.metrics(fadata, overall_score_func=True, if_save=True, if_impute=False)
benchmark_df
```

benchmark visualization

```python
cafe.benchmark.visualize(benchmark_df, add_method_meta=True, save=f"{img_dir}/benchmark_funkeyheatmap.svg") # sorted by wrapper type and overall_score
```

```python
model_name_list = ["ref"] + benchmark_df.index.tolist()
cafe.benchmark.show_trajectory_result_grid(fadata, model_name_list, save=f"{img_dir}/benchmark_trajectory_grid.pdf")
```
