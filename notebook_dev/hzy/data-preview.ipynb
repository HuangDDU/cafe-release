```python
import cafe

cafe.logger.setLevel("DEBUG")
```

```python
dataset_interface_dict = {
    "pancreas": cafe.data.read_pancreas,
    "erythroid_lineage": cafe.data.read_erythroid_lineage,
    "pancrease_cellrank": cafe.data.read_pancrease_cellrank,
    "bonemarrow": cafe.data.read_bonemarrow,
    "hematopoiesis": cafe.data.read_hematopoiesis,
}


figsize = (6, 4)
for k, dataset_interface in dataset_interface_dict.items():
    fadata = dataset_interface()
    print(k, fadata.shape)
    cafe.plot.plot_trajectory(fadata, title="", figsize=figsize)
    cafe.plot.plot_graph(fadata, title="", figsize=figsize, legend_loc=None)
```
