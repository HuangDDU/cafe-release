```python
import cafe
```

data

```python
fadata = cafe.data.read_pancreas()
fadata
```

plot stream

```python
fig_size = (8, 6)
figs = cafe.plot.plot_stream(
    fadata, 
    color="clusters",
    mode="cell",
    fig_size=fig_size,
)

figs = cafe.plot.plot_stream(
    fadata, 
    color="clusters",
    mode="density",
    fig_size=fig_size,
)
```

使用scVelo结果更加好看

```python
model_name = "scvelo"

fadata.load_trajectory_dict(model_name)

figs = cafe.plot.plot_stream(
    fadata, 
    model_name=model_name,
    color="clusters",
    mode="cell",
    fig_size=fig_size,
)

figs = cafe.plot.plot_stream(
    fadata, 
    model_name=model_name,
    color="clusters",
    mode="density",
    fig_size=fig_size,
    
)
```

拆分谱系后识别到驱动基因表达

```python
figs = cafe.plot.plot_stream(
    fadata, 
    model_name=model_name,
    color="Pdx1",
    mode="cell",
    save=True,
    fig_size=fig_size,
)
figs = cafe.plot.plot_stream(
    fadata, 
    model_name=model_name,
    color="Pdx1",
    mode="density",
    save=True,
    fig_size=fig_size,
)
```
